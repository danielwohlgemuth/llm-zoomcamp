```python
import json

from database import Document, Description
from langchain_text_splitters import MarkdownTextSplitter
```

```python
file_name = "docs/translated/Airlock Security Breacher Audit.md"
with open(file_name) as file:
    readme = file.read()
```

```python
splitter_512 = MarkdownTextSplitter(
    chunk_overlap=0,
    chunk_size=512,
)
documents_512 = splitter_512.split_text(readme)
documents_512[:5]
```

```python
splitter_1024 = MarkdownTextSplitter(
    chunk_overlap=0,
    chunk_size=1024,
)
documents_1024 = splitter_1024.split_text(readme)
documents_1024[:5]
```

```python
chunk_id = {
    "file_name": "docs/translated/Airlock Security Breacher Audit.md",
    "chunk_size": 1024,
    "index": 0,
}
json.dumps(chunk_id)
```

```python
database = Document()
```

```python
for index, chunk in enumerate(documents_1024[:5]):
    database.insert(file_name, 1024, index, chunk)
```

```python
with database.connection.cursor() as cursor:
    cursor.execute(
        "SELECT id, file_name, chunk_size, chunk_index, content FROM document"
    )
    records = [row for row in cursor]
    print(f"records: {len(records)}")
    print(records)
```
